# CodeAlpha Internship — Task 2: Exploratory Data Analysis
This notebook explores the synthetic retail sales dataset. The dataset is for practice and is not real company data.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

df = pd.read_csv('data/retail_sales_synthetic.csv', parse_dates=['order_date'])
df.head()
```

## 1. Understand the dataset

```python
print('Shape:', df.shape)
df.info()
df.describe(include='all')
```

## 2. Data quality checks

```python
print('Missing values:\n', df.isna().sum())
print('Duplicate order IDs:', df.order_id.duplicated().sum())
```

## 3. Explore sales by category and region

```python
display(df.groupby('category').sales.agg(['count','sum','mean']).sort_values('sum', ascending=False))
display(df.groupby('region').sales.sum().sort_values(ascending=False))
```

## 4. Monthly trend

```python
monthly = df.assign(month=df.order_date.dt.to_period('M').dt.to_timestamp()).groupby('month', as_index=False).sales.sum()
plt.figure(figsize=(12,5)); plt.plot(monthly.month, monthly.sales, marker='o'); plt.title('Monthly Sales Trend'); plt.xticks(rotation=45); plt.ylabel('Sales'); plt.tight_layout(); plt.show()
```

## 5. Sales and profit relationship

```python
plt.figure(figsize=(7,5)); plt.scatter(df.sales, df.profit, alpha=.45); plt.xlabel('Sales'); plt.ylabel('Profit'); plt.title('Sales vs Profit'); plt.show()
```
